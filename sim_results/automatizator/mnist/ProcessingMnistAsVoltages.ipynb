```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml

# 1. Fetch the MNIST dataset
print("Fetching MNIST data...")
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist.data, mnist.target

# Optional step to mirror starting from a 28x28 matrix:
# X_matrices = X.reshape(-1, 28, 28)
# flattened_data = X_matrices.reshape(-1, 784)

# 2. Explicitly ensure each sample is a flat row of 784 columns
n_samples = X.shape[0]   # 70,000 samples
n_features = X.shape[1]  # 784 columns
flattened_data = X.reshape(n_samples, n_features)

# 3. Normalize the pixel values from [0, 255] to [0, 1.8]
flattened_data = (flattened_data / 255.0) * 1.8


# 4. Create the incremental column: starts at 0, increases by 0.00005 per sample
# This generates: [0.00000, 0.00005, 0.00010, 0.00015, ..., etc.]
index_increment = np.arange(n_samples) * 0.00005

# 5. Construct the Pandas DataFrame
# We create names for pixel columns (0 to 783)
pixel_column_names = [f"pixel_{i}" for i in range(n_features)]
df = pd.DataFrame(flattened_data, columns=pixel_column_names)

# 6. Insert the custom incremental column at the very beginning (index 0)
df.insert(0, 'index_increment', index_increment)

# 7. Format the increment column to preserve float representation in the CSV string
df['index_increment'] = df['index_increment'].map(lambda x: f"{x:.5f}")

# 8. Export the structured matrix to a .csv file
output_file = "mnist_manipulated_flat.txt"
print(f"Exporting to {output_file}...")
df.to_csv(output_file, sep=' ', index=False)

print("Export complete!")
```
