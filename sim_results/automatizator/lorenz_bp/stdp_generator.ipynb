```python
def generar_spice(i_max, j_max, nombre_archivo="red_stdp.spice"):
    contador = 1
    lineas = []

    # Añadimos una cabecera para el archivo SPICE
    lineas.append(f"* Archivo SPICE generado automaticamente ({i_max}x{j_max})")
    
    # Bucles anidados para seguir el patrón fila-columna
    for i in range(1, i_max + 1):
        for j in range(1, j_max + 1):
            # Construcción de la línea siguiendo tu formato exacto
            linea = (
                f"xstdp{contador} avdd avss "
                f"vout_pre_{i} vout_post_{j} iext{j} "
                f"n_vout_pre_{i} n_vout_post_{j} stdp"
            )
            lineas.append(linea)
            contador += 1

    # Guardar las líneas en el archivo .spice
    with open(nombre_archivo, "w") as archivo:
        archivo.write("\n".join(lineas) + "\n")
    
    print(f"¡Archivo '{nombre_archivo}' generado con éxito con {contador - 1} instancias!")

# Ejecutar para i=4, j=2
# generar_spice(i_max=4, j_max=2)
```
